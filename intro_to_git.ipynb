# Introduction to Git

**Git** is a distributed version-control software initially conceived by Linus Torvalds for the development of Linux kernel. 

**Github** is a cloud-based platform to store and manage git repositories. It allows team collaboration.

## Basic Commands

`git init`
- initializes a new Git repository in your directory.

`git clone [repo_url]`
- creates a local copy of a remote repository.

`git status`
- displays the current status of your working directory and staging area.

`git log`
- shows commit history for the repository.

`git add [file]`
- adds changes from your working directory to the staging area.

`git commit -m`
- saves staged changes with a descriptive message.

`git branch`
- default: shows all existing branches
- `-M` renames the actual branch with the input name (ex: git branch -M main)

`git push`
- sends local commits to a remote repository

`git pull`
- sends remote commits to a local repository

`git fetch`
- retrieves updates from remote repositories without merging them.

`git checkout [branch_name]`
- switches to another branch or commit.

`git checkout -b [branch_name]`
- Creates a new branch and switches to it.

`git merge [branch_name]`
- merges the specified branch into the current branch.

## Best Practices

### .gitgnore file

Add sensitive information or large files to the `.gitignore` file. (examples: out-of-memory datasets, API Keys, SSH Keys)

### clear commit's comments

Try to be clear and concise on your commits so you know what and why changes have been made. Avoid large commits.
